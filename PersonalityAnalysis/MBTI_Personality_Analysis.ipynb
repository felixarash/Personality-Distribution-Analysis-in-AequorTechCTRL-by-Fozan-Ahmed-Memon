# MBTI Personality Analysis — AequorTech CTRL Volunteers
Research-style EDA on the volunteering participant dataset created from provided entries.

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

DATA_PATH = os.path.join(os.path.dirname(__file__), 'participants_mbti.csv')
df = pd.read_csv(DATA_PATH)
df.insert(0, 'Participant ID', range(1, len(df)+1))
df['MBTI'] = df['MBTI'].str.upper().str.strip()
df.head()
```

## Distribution of MBTI types
We compute counts and percentages by MBTI and visualize.

```python
mbti_order = [
    'INFP','ENFP','ESFJ','ISTP','INFJ','ENTP','ENTJ','ISFP',
    'INTP','ESTP','ESTJ','ISFJ','ISTJ','ENFJ','INTJ','ESFP'
]
counts = df['MBTI'].value_counts()
percentages = (counts/len(df))*100
summary_df = (
    pd.DataFrame({'MBTI': counts.index, 'Count': counts.values, 'Percentage': percentages.values})
    .set_index('MBTI')
    .reindex(mbti_order)
    .dropna()
    .reset_index()
)
summary_df
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(12,8))
sns.barplot(x='MBTI', y='Percentage', data=summary_df, palette='viridis')
plt.title('Personality Distribution (Volunteering Participants) — AequorTech CTRL')
plt.xlabel('MBTI Type')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=45)
plt.tight_layout()
out_path = os.path.join(os.path.dirname(__file__), 'notebook_participants_personality_distribution.png')
plt.savefig(out_path)
out_path
```
